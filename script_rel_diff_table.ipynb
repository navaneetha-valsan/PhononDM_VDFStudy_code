```python
# Define a function to find the relative difference of rate with respect to the fiducial rate as (Rate_max-Rate_min)/Rate_fid
def relative_difference(rate_max, rate_min, rate_fid):
    if rate_fid == 0:
        return 0
    return (rate_max - rate_min) / rate_fid

# Compute relative difference for a specific mass and (mediator, model, threshold)
def compute_relative_difference(group_max_rate, group_min_rate, group_max_sigma, group_min_sigma, 
                                fiducial_data, mass_value, mediator, model, threshold, rtol):
    """
    Compute the relative difference in rate and sigma at a specific DM mass for a given (mediator, model, threshold).

    Parameters:
        group_max_rate (dict), group_min_rate (dict), group_max_sigma (dict), group_min_sigma (dict): extremes dictionaries
        fiducial_data (dict): Output from read_fiducial_data
        mass_value (float): DM mass value (same units as in data, e.g., eV)
        mediator (str): Mediator name
        model (str): Model name used in group keys
        threshold (float or int): Threshold
        rtol (float): Relative tolerance to identify the mass match

    Returns:
        Tuple: (relative_rate_difference, relative_sigma_difference)
    """
    key = (mediator, model, threshold)

    # Get mass array
    mass_array = np.array(fiducial_data[mediator][threshold]["dm_mass"])

    # Find closest index using tolerance
    diff = np.abs((mass_array - mass_value) / mass_value)
    if np.min(diff) > rtol:
        raise ValueError(f"Mass {mass_value} not found in fiducial data for mediator {mediator}, threshold {threshold}")
    idx = np.argmin(diff)

    rate_fid = fiducial_data[mediator][threshold]["rates"][idx]
    sigma_fid = fiducial_data[mediator][threshold]["sigmas"][idx]

    rate_max = group_max_rate[key][idx]
    rate_min = group_min_rate[key][idx]
    sigma_max = group_max_sigma[key][idx]
    sigma_min = group_min_sigma[key][idx]

    rel_diff_rate = relative_difference(rate_max, rate_min, rate_fid)
    rel_diff_sigma = relative_difference(sigma_max, sigma_min, sigma_fid)

    return rel_diff_rate, rel_diff_sigma, idx
```
